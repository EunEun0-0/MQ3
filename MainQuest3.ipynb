```python
data['CohortQuarter'] = data.groupby('Customer ID')['InvoiceDate'].transform('min').dt.to_period('Q')
```

```python
data
```

```python
data['PurchaseQuarter'] = data['InvoiceDate'].dt.to_period('Q')
```

```python
data
```

```python
def get_cohort_index(df):
    year_diff = df['PurchaseQuarter'].dt.year - df['CohortQuarter'].dt.year
    quarter_diff = df['PurchaseQuarter'].dt.quarter - df['CohortQuarter'].dt.quarter
    return year_diff * 4 + quarter_diff + 1

data['CohortIndex'] = get_cohort_index(data)
```

```python
data
```

```python
cohort_counts = data.groupby(['CohortQuarter', 'CohortIndex'])['Customer ID'].nunique().unstack(1)
```

```python
cohort_counts
```

```python
cohort_sizes = cohort_counts.iloc[:,0]

retention = cohort_counts.divide(cohort_sizes, axis=0) * 100
```

```python
retention
```

```python
#히트맵 코드
heatmap = sns.heatmap(
    data=retention,
    annot=True,                # 셀 내부에 값 표시
    fmt='.2f',                 # 텍스트 형식 (소수점 2자리)
    cmap='Greens',             # 색상 팔레트
    cbar_kws={'label': 'Retention Rate, %'},  # 컬러바 제목
    linewidths=0.5,            # 셀 간격
    linecolor='gray',          # 셀 경계 색상
    vmin=0, vmax=100           # Retention Rate의 범위 설정
)
```

```python
cohort_quantity = data.groupby(['CohortQuarter', 'CohortIndex'])['Quantity'].sum().unstack(1)
```

```python
cohort_customers = data.groupby(['CohortQuarter', 'CohortIndex'])['Customer ID'].nunique().unstack(1)
```

```python
average_quantity = cohort_quantity / cohort_customers
```

```python
heatmap = sns.heatmap(
    data=average_quantity,
    annot=True,                # 셀 내부에 값 표시
    fmt='.1f',                 # 텍스트 형식 (소수점 1자리)
    cmap='Blues',              # 색상 팔레트
    cbar_kws={'label': 'Average Quantity'},  # 컬러바 제목
    linewidths=0.5,            # 셀 간격
    linecolor='gray',          # 셀 경계 색상
    vmin=0                     # 최소값 설정 (필요 시 조정 가능)
)
```

```python
data['InvoiceMonth'] = data['InvoiceDate'].dt.to_period('M')
monthly_revenue_per_user = data.groupby(['InvoiceMonth', 'Customer ID'])['Total'].sum().reset_index()
```

```python
total_revenue_per_month = monthly_revenue_per_user.groupby('InvoiceMonth')['Total'].sum()
```

```python
active_users_per_month = monthly_revenue_per_user.groupby('InvoiceMonth')['Customer ID'].nunique()
```

```python
arpu = total_revenue_per_month / active_users_per_month
```

```python
plt.figure(figsize=(12, 6))
arpu.plot(kind='line', marker='o', color='skyblue')
plt.title("Monthly ARPU (Average Revenue Per User)")
plt.xlabel("Year-Month")
plt.ylabel("ARPU (£)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
customer_df = data.groupby('Customer ID').agg({'Invoice': 'nunique','Total': 'sum'}).reset_index()
customer_df.rename(columns={'Invoice': 'TotalOrders','Total': 'TotalRevenue'}, inplace=True)
customer_df['AOV'] = customer_df['TotalRevenue'] / customer_df['TotalOrders']
purchase_frequency = customer_df['TotalOrders'].sum() / customer_df.shape[0]
```

```python
customer_lifetime = data.groupby('Customer ID').agg({
    'InvoiceDate': ['min', 'max'],
})
customer_lifetime.columns = ['FirstPurchaseDate', 'LastPurchaseDate']
customer_lifetime = customer_lifetime.reset_index()
```

```python
customer_lifetime
```

```python
customer_lifetime['LifetimeDays'] = (customer_lifetime['LastPurchaseDate'] - customer_lifetime['FirstPurchaseDate']).dt.days
```

```python
customer_lifetime
```

```python
customer_lifetime['LifetimeYears'] = customer_lifetime['LifetimeDays'] / 365
```

```python
customer_lifetime
```

```python
customer_df['CLV'] = customer_df['AOV'] * purchase_frequency * customer_lifetime['LifetimeYears']
customer_df = customer_df.sort_values(by='CLV', ascending=False).reset_index(drop=True)
```

```python
np.random.seed(42)
successes = np.random.binomial(n=1, p=0.3, size=10)
print("각 성공 여부 (1: 성공, 0: 실패):", successes)
print("총 성공 횟수:", successes.sum())
```

```python
np.random.seed(42)
scores = np.random.binomial(n=5, p=0.7, size=20)
print("각 학생 점수:", scores)
average_score = scores.mean()
print("전체 평균 점수:", round(average_score, 2))
```

```python
from scipy.stats import norm

np.random.seed(42)
weights = np.random.normal(loc=50, scale=5, size=1000)
within_range = np.mean((weights >= 45) & (weights <= 55))
print(f"45g 이상 55g 이하 비율: {within_range:.2%}")

x = np.linspace(min(weights), max(weights), 1000)
pdf = norm.pdf(x, loc=50, scale=5)
plt.plot(x, pdf, 'r-', linewidth=2)


count, bins, ignored = plt.hist(weights, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7, label='Weight Distribution')
plt.title('Product Weight Distribution')
plt.xlabel('weight (g)')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```
